```python
import f3dasm
import numpy as np
import pandas as pd
```

```python
DIMENSIONALITY = 10000
SEED= 42
```

```python
def get_features(f):
    l = []
    if f.noise == 0.0: l.append('noiseless')
    if f.noise > 0.0: l.append('noisy')
    if (f.dimensionality <= 5): l.append('low_dimensional')
    if (5 < f.dimensionality <= 10): l.append('medium_dimensional')
    if (f.dimensionality >= 10): l.append('high_dimensional')
    if f.convex: l.append('convex')
    if f.separable: l.append('seperable')
    if f.differentiable: l.append('differentiable')
    if f.multimodal: l.append('multimodal')
    if f.randomized_term: l.append('randomized_term')
    if f.parametric: l.append('parametric')
    return ",".join(l)
```

```python
l = []
number_of_functions = 1000
```

```python
dim = np.random.choice([2, 10, 20, 1000])
type(dim)
```

```python
for i in range(number_of_functions):
    dim = np.random.choice([2, 10, 20, 1000])
    noise = np.random.choice([0.0, 0.1, 0.2, 0.4])
    f = np.random.choice(f3dasm.functions.get_functions(d=int(dim)))
    func = f(dimensionality=dim, noise=noise)
    l.append(get_features(func))

df_features = pd.DataFrame(l)
```
