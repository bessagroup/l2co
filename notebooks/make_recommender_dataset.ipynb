````python
import pandas as pd
import numpy as np
import random
import datetime
from recommenders.utils.constants import DEFAULT_HEADER
import os
import glob
````

````python
hash: int = random.getrandbits(32)
````

````python
ts: float = datetime.datetime.now().timestamp()
````

````python

df = pd.read_csv ('./filename.csv', index_col=0)
df
````

````python
# Take the last row; end of optimization process
scores = df.iloc[-1].values
scores
````

````python
min_scores = np.min(scores)
max_scores = np.max(scores)

L = (scores - min_scores)/(max_scores - min_scores)
L
````

````python
rating_array = (4*L)+1
````


````python
data = [[i+1, hash, f"{rating:.5f}", f"{ts:.0f}"] for i, rating in enumerate(rating_array)]


data
df = pd.DataFrame(data, columns=DEFAULT_HEADER)
df.to_csv('g.csv', index=False)
````
